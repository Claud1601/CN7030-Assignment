```python
import pandas as pd

# Load the property dataset
property_df = pd.read_csv('/content/property.csv')

# Display the first 5 rows of the DataFrame
display(property_df.head())
```

```python
# Display basic information about the DataFrame to check for missing values and data types
property_df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np

# Define the dependent variable (target)
target = 'Price'
y = property_df[target]

# Define the potential independent variables (features)
all_features = ['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms', 'Year_Built', 'Lot_Size']

# Split the data into training and testing sets
# Using a random state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(property_df[all_features], y, test_size=0.2, random_state=42)
```

```python
# Define different combinations of features for multiple linear regression models
feature_combinations = {
    "Model 1: Square_Footage, Num_Bedrooms, Num_Bathrooms":
        ['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms'],
    "Model 2: Square_Footage, Num_Bedrooms, Num_Bathrooms, Year_Built":
        ['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms', 'Year_Built'],
    "Model 3: Square_Footage, Num_Bedrooms, Num_Bathrooms, Year_Built, Lot_Size":
        ['Square_Footage', 'Num_Bedrooms', 'Num_Bathrooms', 'Year_Built', 'Lot_Size'],
    "Model 4: Square_Footage, Year_Built, Lot_Size":
        ['Square_Footage', 'Year_Built', 'Lot_Size'],
    "Model 5: Num_Bedrooms, Num_Bathrooms, Year_Built, Lot_Size":
        ['Num_Bedrooms', 'Num_Bathrooms', 'Year_Built', 'Lot_Size']
}

# Dictionary to store R-squared scores for each model
r_squared_scores = {}

# Build and evaluate models
for model_name, features in feature_combinations.items():
    print(f"\n--- Building {model_name} ---")

    # Select features for the current model
    X_train_model = X_train[features]
    X_test_model = X_test[features]

    # Initialize and train the Linear Regression model
    model = LinearRegression()
    model.fit(X_train_model, y_train)

    # Make predictions on the test set
    y_pred = model.predict(X_test_model)

    # Calculate R-squared score
    r2 = r2_score(y_test, y_pred)
    r_squared_scores[model_name] = r2

    print(f"Features used: {', '.join(features)}")
    print(f"R-squared: {r2:.4f}")

# Display all R-squared scores for comparison
print("\n--- R-squared Scores Comparison ---")
for model_name, r2 in r_squared_scores.items():
    print(f"{model_name}: {r2:.4f}")
```
